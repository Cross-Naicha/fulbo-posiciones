```python
class Reunion:

    def __init__(self, field, food, guests):
        
        self.field = field
        self.food = food
        self.guests = guests
        
        self.foodies = [guest for guest in guests if guest.foodie == True]
        self.players = [guest for guest in guests if guest.player == True]

        self.foodies_n = len(self.foodies)
        self.players_n = len(self.players)

        self.food_fraction = self.food / len(self.foodies)
        self.field_fraction = self.field / len(self.players)

        for foodie in self.foodies:
            foodie.food_payment = self.food_fraction

class Guest:

    def __init__(self, name, player, foodie):
        self.name = name
        self.player = player
        self.foodie = foodie

        self.food_payment = 0
        self.field_payment = 0
```

```python
def boolenizer(value):
    return True if value == '1' else False
```

```python
guests = "mario chicho naicha benja recalde juan lomba pollo cesar emi bag ponce".split(" ")
GUESTS = []

for guest in guests:
    
    player = input(f"{guest} juega?")
    foodie = input(f"{guest} come?")

    GUESTS.append(Guest(guest, boolenizer(player), boolenizer(foodie)))
```
